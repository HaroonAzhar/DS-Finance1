```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# dataset url = https://www.kaggle.com/datasets/mlg-ulb/creditcardfraud/data

data = '../data/raw/creditcard.csv'

df = pd.read_csv(data)


# Show basic info
print("df dimention: ", df.shape)
print("df class distribution: ", df["Class"].value_counts()) # Class distribution
df.info()
```

```python
# Get feature names
target = ['Class']
feature_names = df.columns.difference(target)

# map class labels for readability
df["Class_Label"] = df["Class"].map({0: "Non-Fraud", 1: "Fraud"}) ## Class – binary label: 0 = genuine ; 1 = fraud

n_cols = 4
n_rows = 8

def create_box_plot(feature_names, x1=30, y1=30, x2=8, y2=4):
    # Helper function to Create boxplots
    plt.figure(figsize=(x1, y1))

    for i, feature in enumerate(feature_names, 0):
        plt.subplot(x2 ,y2, i+1)
        sns.boxplot(x="Class_Label", y=feature, data=df)
        plt.title(f"{feature} by Class")
        plt.xlabel("")
        plt.ylabel(feature)

    plt.tight_layout()
    plt.show()

create_box_plot(feature_names)
```

```python
# box plot of most important features to visualize relationship between distribution of variables and class categories.

most_imp_feat = ['V14', 'V10', 'V12', 'V4', 'V17', 'Amount']
create_box_plot(most_imp_feat, 12, 8, 3, 2) # Passing feature list and plot/subplot dimentions.
```

## From the Box Whisker plots above, we could see 
### 1. Most Fradulant transactions falls as outlier of Features V17, V14, V10, V12 & V4. Note: The commercial features have been masked in the dataset so its not descriptive.
### 2. Fradulant transactions occurs in lower amounts

```python
df = df.drop(['Class_Label'], axis=1)
correlation_matrix = df.corr(method='pearson').round(1)
print(correlation_matrix)
```

```python

# Assuming you want correlations relative to a specific target variable, e.g., 'target_col'
target_correlation = correlation_matrix['Class'].sort_values(ascending=False)

# Filter for strong correlations (absolute value > 0.5)
impactful_variables = target_correlation[abs(target_correlation) > 0.1]

print("Impactful variables based on a |r| > 0.1 threshold:")
print(impactful_variables)
```
